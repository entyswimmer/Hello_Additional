```python
def battle(atk, hp, pp):
    hp = hp - atk
    pp = pp -1
    return hp, pp
```

```python
liz = {"name": "リザードン", "hp": 400, "atk": 220, "skill": "ダイモンジ", "pp": 5}
fus = {"name": "フシギバナ", "hp": 300, "atk": 120, "skill": "ソーラービーム", "pp": 5}
```

```python
def main(m1, m2):
    print(f"{m1["name"]}の{m1["skill"]}!!")
    if m1["pp"] > 0 and m2["hp"] > m1["atk"]:
        new_hp, new_pp = battle(m1["atk"], m2["hp"], m1["pp"])
        m2["hp"] = new_hp
        m1["pp"] = new_pp
        print(f"{m2["name"]}に{m1["atk"]}ダメージ")
        print(f"{m2["name"]}のHPは{max(0, m2["hp"])}になった!")
    elif m1["pp"] <= 0:
        print(f"{m1["name"]}のPPが足りない!")
    else:
        print(f"{m2["name"]}は倒れた!")
```

```python
main(liz, fus)
```
