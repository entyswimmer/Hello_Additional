## ・マルウェアのダミー作成

今回作成するマルウェアはランサムウェアという、データを暗号化してしまうマルウェアです。

・マルウェアとウイルスの違い<br>
    マルウェア: 悪意のあるソフトウェア全般を指す<br>
    ウイルス: マルウェアの一種で、他のプログラムやファイルに感染していく<br>

```python
#演習問題１: 攻撃コードを作成してみよう
"""
絶対に悪用しないこと(このコードでは本番環境に使用できません)
"""

def ransomware(data):
    key = os.urandom(32)
    new_data = []
    new_file = []
    
    i = 0
    for file in data:
        progress = (i+1)/len(file)
        sys.stdout.write(f'\rProcessing:[{int(progress * 20) * "="}> {int((1-progress) * 20) * " "}] {progress:.1%}')
        sys.stdout.flush()
        for info in file:
            cipher_aes = AES.new(key, AES.MODE_GCM)
            c, _ = cipher_aes.encrypt_and_digest(info.encode('utf-8'))
            new_file.append(c)
        new_data.append(new_file)
        i = i + 1
        time.sleep(3)
    
    sys.stdout.write('\n侵入完了\n')
    print("======================================\n")
    
    i = 0
    for file in new_data:
        print(f"・file{i}の情報")
        print(file, "\n")
        i = i+1

    print("=================warning!!======================\n")
    print("1分以内に1000万円を振り込まないと、データを破棄します\n")
    print("================================================\n")

    def countdown_timer(duration_seconds):
    
        timer_display = display(HTML("<h2>残り時間: 00:00</h2>"), display_id=True)
        
        start_time = time.time()
        end_time = start_time + duration_seconds
    
        while time.time() < end_time:
            remaining_seconds = int(end_time - time.time())
            time_format = str(datetime.timedelta(seconds=remaining_seconds))[2:]
            html_content = f"<h2>残り時間: {time_format}</h2>"
            timer_display.update(HTML(html_content))
            time.sleep(1)
    
        timer_display.update(HTML("<h2>1分経過しました。共有鍵は削除します。</h2>"))

    countdown_timer(60)
```

```python
ransomware(data)
```

マルウェアを検知して、上のランサムウェアが実行されるのを防ごう<br>

マルウェアは実行ファイルのハッシュ値のデータベースがあり、それを元に防御する<br>
-> 今回作成したマルウェア(今回は関数)のハッシュ値を確認してみよう

```python
hash(ransomware)
```

```python
#出力されたハッシュ値を用いて, マルウェア関数の検知を行おう
def detect(func):
    if hash(func) == 286457266:
        print(f"ランサムウェアを{func}から検出しました！")
        print("今すぐ該当ファイルを削除してください!")
    else:
        print("ファイル")
```

```python
detect(ransomware)
```
